```python
#@title Run Generator
API_ID = "" #@param {type:"string"}
API_HASH = "" #@param {type:"string"}

import asyncio
from pyrogram import Client

async def main():
    async with Client(':memory:', api_id=int(API_ID.strip()), api_hash=API_HASH.strip()) as app:
        print('\nYour USER_SESSION_STRING is:\n')
        print(await app.export_session_string())

try:
    import nest_asyncio
    nest_asyncio.apply()
    asyncio.get_event_loop().run_until_complete(main())
except Exception as e:
    import traceback
    traceback.print_exc()
```
